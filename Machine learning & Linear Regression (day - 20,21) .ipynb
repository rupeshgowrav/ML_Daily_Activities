# Machine learning Introduction 

#### What is Machine Learning? 
Machine learning is a method where a computer learns patterns from past data and uses those patterns to make predictions or decisions on new data, without being explicitly programmed for every rule.


## Traditional Programming vs Machine learning
#### Traditional programming :
   It is a rule-based approach where humans write fixed instructions and the computer follows those rules to produce output
   - Formula : Rules + Data -> Output
#### Machine Learning:
   It is needed when writing rules manually is impossible, if data is very large and patterns keep changing over time
   - Formula: Data + Output -> Rules

#### Dataset 
A dataset is a collection of past observations stored in tabular or structured form, used to train a machine learning model 
- Row -> Data points
- Columns -> Features

## Types of machine learning


#### Supervised machine learning
Supervised learning is a type of machine learning where the model is trained using labelled data , meaning both input and correct output are known.


#### Unsupervised machine learning
Unsupervised learning is a type of machine learning where the model is trained using unlabelled data and discovers hidden patterns or groupd by itself.
                    

#### Reinforcement Learning
Reinforcement Learning is a type of machine learning where an agent learns by interacting with an environment, takes actions and improves its behavior over time by receiving rewards for correct actions and penalities for wrong actions.

# 1. Supervised learning 
## Classification :
Classification in machine learning is a supervised learning technique used to predict discrete class labels(countable) or categories based on input features
### Key concepts - Discrete output:
- The output of a classification model belongs to a fixed set of categories such as : Yes/No , Spam / Not spam.
- Classification models are trained using labelled datasets, where each data points includes: input features(X) , Correct Class Label(Y).
- The model learns a decision boundary that separates different classes in the feature space.
- This boundary can be a line, curve , or complex shape, depending on the algorithm.
### Common Classification Algorithms: 
- Logistic Regression
- K-Nearest Neighbors(KNN)
- Naive Bayes
- Decision Tree
- Random Forest
- Support Vector Machine(SVM)
### Applications:
- Email filterning: Spam detection
- Healthcare: Disease diagnosis
- Finance: Fraud Detection, credit scoring
- Computer vision: Face recognition, object detection
- Marketing: Customer segmentation, churn prediction
### Independent and dependent variables: 
- independent variables(features): Inputs used for prediction
- Dependent variables(target): The class label being predicted
## Types of Classification:
 1. Binary Classification : 
- Two classes
- Example: pass/Fail , Spam/Not Spam
 2. Multiclass classification: 
- More than two classes
- Example: Digit recognition(0-9), Grade(A/B/C/D)

# Regression
#### What is Regression?
- Regression in machine learning is a supervised learning technique used to model relationships between input variables(features) and a continuous output variable(target), with the goal of predicting numerical outcomes.
- It is distinct from classification, which predicts discrete categories or class labels.
#### Key Concepts - continuous output: 
- Output: The primary characteristic of regression is that its output can be any numerical value within a range(e.g, price, temperature, age), rather than a fixed class(e.g, yes/no , spam/not spam)
- Best-fit Line/Curve: The model aims to find a line or curve that best fits the data points, minimizing the difference(error or residual) between the actual value and predicted value.

#### Common types of regression algorithms : 
- Linear Regression
- Polynomial Regression
- Ridge Regression
- Lasso Regression
#### Applications Regression:
- Predicting stock prices in finance.
- Estimating property values in Real Estate.
- Forecasting sales in marketing & sales.
- Disease risks in Healthcare.

# Linear Regression 


#### What is linear regression 
- Linear regression is a supervised learning algorithm used to predict continuous numerical values.
- It models the relationship between a dependent variable(Y) and one or more independent variables(X) using a straight line or plane.
- Example: Predicting marks based on study hours, predicting house prices based on area.


### types of linear regression :
 1. simple linear regression
 2. Multiple linear regression

### Simple linear Regression 
- Uses only one independent variable (X) to predict a dependent variable(Y).
- The goal is to find a straight line that best fits the data points.
- 

### Multiple Linear Regression
- uses more than one independent variable(X1,X2,...,Xn) to predict Y.
- The goal is to find a hyperplane that best fits the data.


### Simple Linear regression vs multiple linear regression


# Simple Linear Regression

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt 
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame({
    "Study_Hours" :[1,2,3,4,5],
    "Marks" :[2,4,5,4,5]
})
```

```python
df
```

```python
x=df[["Study_Hours"]]
y=df[["Marks"]]
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
x_line = pd.DataFrame(
    {'Study_Hours': np.linspace(x.min()[0],x.max()[0],100)}
)
y_line = model.predict(x_line)
```

```python
print("Slope (m):", model.coef_[0])
print("Intercept (b):", model.intercept_)
```

```python
x_new = pd.DataFrame({'Study_Hours': [4.0]})
y_new = model.predict(x_new)
```

```python
print("Predict Value:",y_new)
```

```python
# plot
plt.figure()
plt.scatter(x,y)                        
plt.plot(x_line,y_line)                 
plt.scatter(x_new,y_new)
plt.xlabel("Study_Hours")
plt.ylabel("Marks")
plt.title("Linear Regression: Best-Fit Line & Prediction")
plt.show()
```

# Finding the Best Fit Line
- Goal: Minimize the Sum of Squarred Errors(SSE)
- Error/Residual = Actual_y-Predicted_y
- Formula(MSE):
- (1/n) * Σ(Residual/Error)



```python
from sklearn.metrics import mean_squared_error

y_pred = model.predict(x)
mse = mean_squared_error(y,y_pred)

print("MSE:", mse)
```

# Multi Linear Regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df=pd.read_csv("MLR_Study_marks.csv")
```

```python
df
```

```python
print("\nDataset Info:")
print(df.info())
```

```python
print(df.describe())
```

```python
x = df[["Study_Hours","Attendance","Sleep_Hours"]]
y = df["Marks"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size =0.2)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)

print("Intercept:", model.intercept_)
print("Slope:", model.coef_)
```

```python
y_pred = model.predict(x_test)
```

```python
plt.scatter(y_test,y_pred)
plt.xlabel("Actual Marks")
plt.ylabel("Predicted Marks")
plt.title("Actual vs Predicted Marks")
plt.plot([y_test.min(), y_test.max()],[y_test.min(),y_test.max()],'r--')
plt.show()
```

```python
mse = mean_squared_error(y_test,y_pred)
print("\nMean Squared Error:", mse)
```

```python
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(df["Study_Hours"],df["Attendance"],df["Marks"],c='b',marker='o')
ax.set_xlabel("Study Hours")
ax.set_ylabel("Attendance")
ax.set_zlabel("Marks")
ax.set_title("3D plot: Study Hours & Attendance vs Marks")
plt.show()
```
